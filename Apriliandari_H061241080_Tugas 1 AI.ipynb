# PENDAHULUAN

Perkembangan Artificial Intelligence (AI) memberikan kemudahan dalam kegiatan akademik dan penelitian, termasuk dalam bidang geofisika. AI dapat membantu dalam pengolahan data, analisis, interpretasi, dan penyusunan informasi. Namun, hasil dari AI juga dipengaruhi oleh prompt yang diberikan. Oleh karena itu, diperlukan prompt yang jelas dan sesuai dengan tujuan yang ingin dicapai.

Dalam tugas ini, prompt disusun menggunakan struktur work specification yang terdiri dari Role (R), Task (T), Context (C), Input (I), Constraints (K), Output (O), Quality Criteria (Q), dan Verification (V). Struktur ini digunakan agar perintah yang diberikan kepada AI memiliki tujuan, data, batasan, dan hasil yang jelas.

Penerapan work specification dilakukan pada lima topik geofisika, yaitu analisis data seismik, anomali gravitasi, data resistivitas untuk eksplorasi air tanah, aktivitas gempa bumi, serta hubungan aktivitas petir dengan parameter atmosfer. Melalui tugas ini, diharapkan dapat dipahami cara membuat prompt yang lebih terarah untuk membantu menyelesaikan berbagai permasalahan dalam bidang geofisika.

# ANALISIS DATA SEISMIK

## Role
Kita berperan sebagai mahasiswa geofisika yang sedang belajar membaca dan memahami data seismik.

## Task
Menganalisis penampang seismik 2D untuk melihat batas lapisan bawah permukaan, mencari reflektor yang terlihat jelas, dan melihat apakah ada kemungkinan struktur geologi pada penampang.

## Context
Data seismik digunakan untuk membantu melihat kondisi bawah permukaan. Gelombang seismik akan dipantulkan ketika bertemu dengan batas antara lapisan yang memiliki sifat fisik berbeda. Pantulan tersebut kemudian dapat dilihat pada penampang seismik dalam bentuk reflektor. Dalam analisis ini, pola reflektor akan diperhatikan dari bentuk, kontinuitas, dan perubahan amplitudonya.

## Input
Data yang digunakan berupa penampang seismik 2D dalam bentuk gambar atau data angka. Data tersebut memiliki informasi waktu tempuh dua arah (*Two-Way Travel Time* atau TWT).

## Constraints
1. Jangan langsung menentukan jenis batuan hanya dari bentuk reflektornya.
2. Gunakan apa yang benar-benar terlihat pada penampang sebagai dasar analisis.
3. Bedakan antara hasil pengamatan dan dugaan.
4. Jangan menambahkan struktur geologi jika tidak terlihat pada data.
5. Jika ada bagian yang kurang jelas, jelaskan bahwa hasilnya masih belum pasti.

## Output
Hasil analisis berisi:
1. Reflektor utama yang terlihat pada penampang.
2. Batas lapisan yang kemungkinan ada.
3. Kemungkinan struktur geologi yang terlihat.
4. Alasan sederhana dari setiap hasil interpretasi.
5. Kesimpulan singkat mengenai kondisi bawah permukaan.

## Quality Criteria
Hasil analisis harus sesuai dengan pola yang terlihat pada penampang seismik. Analisis dilakukan dengan memperhatikan bentuk reflektor, kontinuitas reflektor, perubahan amplitudo, dan posisi TWT. Hasilnya harus menggunakan bahasa yang sederhana dan mudah dipahami. Jika suatu interpretasi masih berupa dugaan, hal tersebut harus dijelaskan.

## Verification
Periksa kembali hasil analisis dengan melihat penampang seismik secara keseluruhan. Pastikan setiap reflektor, batas lapisan, dan kemungkinan struktur yang disebutkan memang memiliki dasar dari data. Jika ada bagian yang tidak jelas, jangan dipaksakan untuk diinterpretasikan dan berikan keterangan bahwa bagian tersebut masih memiliki ketidakpastian.

# INTERPRETASI ANOMALI GRAVITASI

## Role
Kita berperan sebagai mahasiswa geofisika yang sedang belajar membaca dan memahami data gravitasi.

## Task
Menganalisis data anomali gravitasi untuk melihat pola anomali, daerah dengan nilai gravitasi tinggi dan rendah, serta melihat apakah ada kemungkinan perubahan kondisi atau struktur bawah permukaan.

## Context
Data gravitasi digunakan untuk membantu mengetahui kondisi bawah permukaan berdasarkan perbedaan densitas material. Perbedaan densitas tersebut dapat menyebabkan nilai anomali gravitasi yang berbeda pada suatu daerah. Dalam analisis ini, pola anomali akan diperhatikan dari nilai, bentuk, dan persebarannya.

## Input
Data yang digunakan berupa data anomali gravitasi dalam bentuk tabel, gambar, atau peta. Data tersebut memiliki informasi lokasi dan nilai anomali gravitasi.

## Constraints
1. Jangan langsung menentukan jenis batuan hanya dari nilai anomali gravitasi.
2. Gunakan pola anomali yang benar-benar terlihat pada data sebagai dasar analisis.
3. Bedakan antara hasil pengamatan dan dugaan.
4. Jangan menambahkan struktur geologi jika tidak memiliki pola anomali yang mendukung.
5. Jika ada bagian yang kurang jelas, jelaskan bahwa hasilnya masih belum pasti.

## Output
Hasil analisis berisi:

1. Pola anomali gravitasi yang terlihat.
2. Daerah dengan anomali gravitasi relatif tinggi dan rendah.
3. Kemungkinan perubahan kondisi atau struktur bawah permukaan.
4. Alasan sederhana dari setiap hasil interpretasi.
5. Kesimpulan singkat mengenai kondisi bawah permukaan.

## Quality Criteria
Hasil analisis harus sesuai dengan pola yang terlihat pada data gravitasi. Analisis dilakukan dengan memperhatikan nilai anomali, pola persebaran, dan perubahan nilai pada daerah tertentu. Hasilnya harus menggunakan bahasa yang sederhana dan mudah dipahami. Jika suatu interpretasi masih berupa dugaan, hal tersebut harus dijelaskan.

## Verification

Periksa kembali hasil analisis dengan melihat data gravitasi secara keseluruhan. Pastikan setiap daerah anomali dan kemungkinan struktur yang disebutkan memang memiliki dasar dari data. Jika ada bagian yang tidak jelas, jangan dipaksakan untuk diinterpretasikan dan berikan keterangan bahwa bagian tersebut masih memiliki ketidakpastian.

# INTERPRETASI DATA RESISTIVITAS UNTUK EKSPLORASI AIR TANAH

## Role

Kita berperan sebagai mahasiswa geofisika yang sedang belajar membaca dan memahami data resistivitas untuk mencari air tanah.

## Task

Menganalisis data resistivitas untuk mengetahui lapisan bawah permukaan dan melihat kemungkinan adanya lapisan yang mengandung air tanah.

## Context

Metode resistivitas digunakan untuk mengetahui kondisi bawah permukaan berdasarkan perbedaan nilai tahanan jenis material. Nilai resistivitas dapat berbeda antara material yang kering, basah, atau memiliki kandungan air. Dalam analisis ini, pola dan nilai resistivitas akan digunakan untuk membantu memperkirakan kondisi lapisan bawah permukaan.

## Input

Data yang digunakan berupa data resistivitas dalam bentuk tabel, grafik, atau penampang 2D. Data tersebut memiliki informasi nilai resistivitas dan kedalaman atau posisi pengukuran.

## Constraints

1. Jangan langsung menentukan jenis batuan hanya dari nilai resistivitas.
2. Jangan langsung menyatakan suatu lapisan sebagai air tanah hanya berdasarkan nilai resistivitas.
3. Gunakan pola dan nilai resistivitas yang terlihat pada data sebagai dasar analisis.
4. Bedakan antara hasil pengamatan dan dugaan.
5. Jika ada bagian yang kurang jelas, jelaskan bahwa hasilnya masih belum pasti.

## Output

Hasil analisis berisi:

1. Lapisan dengan nilai resistivitas yang berbeda.
2. Perkiraan batas antar lapisan.
3. Kemungkinan lapisan yang mengandung air tanah.
4. Alasan sederhana dari setiap hasil interpretasi.
5. Kesimpulan singkat mengenai kondisi bawah permukaan.

## Quality Criteria

Hasil analisis harus sesuai dengan nilai dan pola resistivitas yang terlihat pada data. Analisis dilakukan dengan memperhatikan perubahan nilai resistivitas, kedalaman, dan bentuk lapisan. Hasilnya harus menggunakan bahasa yang sederhana dan mudah dipahami. Jika suatu lapisan diduga mengandung air tanah, hal tersebut harus dijelaskan sebagai dugaan.

## Verification

Periksa kembali hasil analisis dengan melihat data resistivitas secara keseluruhan. Pastikan setiap lapisan dan kemungkinan keberadaan air tanah yang disebutkan memiliki dasar dari data. Jika ada bagian yang tidak jelas, jangan dipaksakan untuk diinterpretasikan dan berikan keterangan bahwa bagian tersebut masih memiliki ketidakpastian.

# ANALISIS AKTIVITAS GEMPA BUMI

## Role

Kita berperan sebagai mahasiswa geofisika yang sedang belajar membaca dan memahami data gempa bumi.

## Task

Menganalisis data gempa bumi untuk melihat persebaran lokasi, waktu kejadian, magnitudo, dan kedalaman gempa pada suatu wilayah.

## Context

Data gempa bumi dapat digunakan untuk mengetahui aktivitas kegempaan pada suatu wilayah. Informasi seperti lokasi, waktu kejadian, magnitudo, dan kedalaman dapat membantu melihat pola persebaran gempa dan daerah yang memiliki aktivitas gempa lebih banyak.

## Input

Data yang digunakan berupa data gempa bumi dalam bentuk tabel atau peta. Data tersebut memiliki informasi seperti waktu kejadian, koordinat, magnitudo, dan kedalaman gempa.

## Constraints

1. Jangan menentukan penyebab gempa hanya berdasarkan satu informasi.
2. Gunakan data yang tersedia sebagai dasar analisis.
3. Bedakan antara hasil pengamatan dan dugaan.
4. Jangan menentukan keberadaan sesar jika tidak ada pola data yang mendukung.
5. Jika ada bagian yang kurang jelas, jelaskan bahwa hasilnya masih belum pasti.

## Output

Hasil analisis berisi:

1. Pola persebaran gempa bumi.
2. Perubahan aktivitas gempa berdasarkan waktu.
3. Persebaran magnitudo dan kedalaman gempa.
4. Kemungkinan hubungan aktivitas gempa dengan kondisi geologi.
5. Kesimpulan singkat mengenai aktivitas gempa pada wilayah tersebut.

## Quality Criteria

Hasil analisis harus sesuai dengan data gempa yang diberikan. Analisis dilakukan dengan memperhatikan lokasi, waktu, magnitudo, dan kedalaman gempa. Hasilnya harus menggunakan bahasa yang sederhana dan mudah dipahami. Jika terdapat dugaan mengenai penyebab atau kondisi geologi, hal tersebut harus dijelaskan sebagai dugaan.

## Verification

Periksa kembali hasil analisis dengan melihat data gempa secara keseluruhan. Pastikan pola persebaran, waktu, magnitudo, dan kedalaman yang disebutkan sesuai dengan data. Jika ada bagian yang tidak jelas, jangan dipaksakan untuk diinterpretasikan dan berikan keterangan bahwa bagian tersebut masih memiliki ketidakpastian.

# ANALISIS HUBUNGAN AKTIVITAS PETIR DENGAN PARAMETER ATMOSFER

## Role

Kita berperan sebagai mahasiswa geofisika yang sedang belajar menganalisis aktivitas petir dan kondisi atmosfer.

## Task

Menganalisis hubungan antara aktivitas petir dengan parameter atmosfer untuk melihat apakah perubahan kondisi atmosfer berhubungan dengan meningkat atau menurunnya aktivitas petir.

## Context

Aktivitas petir dapat dipengaruhi oleh kondisi atmosfer. Beberapa parameter atmosfer seperti suhu, kelembapan, tekanan udara, atau CAPE dapat digunakan untuk melihat kondisi atmosfer saat terjadi petir. Dalam analisis ini, perubahan aktivitas petir akan dibandingkan dengan perubahan parameter atmosfer yang digunakan.

## Input

Data yang digunakan berupa data aktivitas petir dan data parameter atmosfer dalam bentuk tabel atau grafik. Data tersebut memiliki informasi waktu dan nilai aktivitas petir serta parameter atmosfer.

## Constraints

1. Jangan langsung menyatakan bahwa suatu parameter atmosfer menyebabkan terjadinya petir.
2. Gunakan data yang tersedia sebagai dasar analisis.
3. Bedakan antara hasil pengamatan dan dugaan.
4. Perhatikan kesesuaian waktu antara data petir dan parameter atmosfer.
5. Jika hubungan yang terlihat lemah atau tidak jelas, jelaskan bahwa hasilnya masih belum pasti.

## Output

Hasil analisis berisi:

1. Pola aktivitas petir berdasarkan waktu.
2. Perubahan nilai parameter atmosfer.
3. Hubungan antara aktivitas petir dengan parameter atmosfer.
4. Alasan sederhana dari hasil analisis.
5. Kesimpulan singkat mengenai hubungan aktivitas petir dengan kondisi atmosfer.

## Quality Criteria

Hasil analisis harus sesuai dengan pola data petir dan parameter atmosfer yang diberikan. Analisis dilakukan dengan memperhatikan perubahan nilai dan kesesuaian waktu antar data. Hasilnya harus menggunakan bahasa yang sederhana dan mudah dipahami. Jika terdapat hubungan yang belum dapat dipastikan, hal tersebut harus dijelaskan.

## Verification

Periksa kembali hasil analisis dengan membandingkan data petir dan parameter atmosfer secara keseluruhan. Pastikan hubungan yang disebutkan memiliki dasar dari data dan tidak hanya berdasarkan kebetulan pada beberapa waktu tertentu. Jika data tidak menunjukkan hubungan yang jelas, nyatakan bahwa hasilnya memiliki ketidakpastian.
